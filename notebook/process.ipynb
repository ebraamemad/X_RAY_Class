```python
import pandas as pd
from glob import glob
import os
from pathlib import Path


def build_data_map(folder_path):
    imgs_paths = glob(os.path.join(folder_path, "*.jpeg"))
    clas = Path(folder_path).name
    split = Path(folder_path).parent.name
    return pd.DataFrame({
        "imgs": imgs_paths,
        "clas": clas,
        "split": split
    })

test_normal = build_data_map(r"E:\projects\Chest X-Ray\load_data\chest_xray\chest_xray\test\NORMAL")
test_pneumonia = build_data_map(r"E:\projects\Chest X-Ray\load_data\chest_xray\chest_xray\test\PNEUMONIA")

train_normal = build_data_map(r"E:\projects\Chest X-Ray\load_data\chest_xray\chest_xray\train\NORMAL")
train_pneumonia = build_data_map(r"E:\projects\Chest X-Ray\load_data\chest_xray\chest_xray\train\PNEUMONIA")

val_normal = build_data_map(r"E:\projects\Chest X-Ray\load_data\chest_xray\chest_xray\val\NORMAL")
val_pneumonia = build_data_map(r"E:\projects\Chest X-Ray\load_data\chest_xray\chest_xray\val\PNEUMONIA")
```
